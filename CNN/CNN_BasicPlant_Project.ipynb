```python
import os

dataset_path = '/content/plant_dataset'

for root, dirs, files in os.walk(dataset_path):
    print("Folder:", root)
    print("Number of files:", len(files))
```

```python
import tensorflow as tf

# Define dataset directory paths based on downloaded structure
TRAIN_DIR = "/content/plant_dataset/New Plant Diseases Dataset(Augmented)/New Plant Diseases Dataset(Augmented)/train"
VALID_DIR = "/content/plant_dataset/New Plant Diseases Dataset(Augmented)/New Plant Diseases Dataset(Augmented)/valid"

IMG_SIZE = (224, 224)
BATCH_SIZE = 32

train_ds = tf.keras.utils.image_dataset_from_directory(
    TRAIN_DIR,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="int",
    shuffle=True,
    seed=42
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    VALID_DIR,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="int",
    shuffle=False
)

class_names = train_ds.class_names
NUM_CLASSES = len(class_names)

print("Number of classes:", NUM_CLASSES)
print("Class names:", class_names)
```

```python
class_names = train_ds.class_names

print(class_names)
```

```python
for images, labels in train_ds.take(1):

    print("Images shape:", images.shape)
    print("Labels:", labels.numpy())
```

```python
import tensorflow
from tensorflow import keras
from keras.layers import Conv2D,Dense,MaxPooling2D,Flatten
from keras.models import Sequential
```

```python
def Normalize(image, label):
    image = tf.cast(image / 255.0, tf.float32)

    return image, label
```

```python
train_ds=train_ds.map(Normalize)
```

```python
model=Sequential()
model.add(Conv2D(32,kernel_size=(3,3),padding='valid',activation='relu',input_shape=(223,224,3)))
model.add(MaxPooling2D(padding='valid',pool_size=(3,3),strides=2))
model.add(Conv2D(32,kernel_size=(3,3),padding='valid',activation='relu'))
model.add(MaxPooling2D(padding='valid',pool_size=(3,3),strides=2))

model.add(Flatten())

model.add(Dense(128,activation='relu'))
model.add(Dense(NUM_CLASSES,activation='softmax'))

model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
val_ds = val_ds.map(Normalize)
```

```python

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=5
)
```

```python

test_loss, test_accuracy = model.evaluate(val_ds)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy * 100, "%")
```
