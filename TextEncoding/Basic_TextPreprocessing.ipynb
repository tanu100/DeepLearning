# **Basic Understanding of Text Preprocessing for Basic ML and DL Before NLP**

**NLTK (Natural Language Toolkit)** is a Python library used for Natural Language Processing (NLP). It provides tools to process, analyze, and manipulate human languages (English, Hindi, etc.) so that machines can understand textual data.

In simple words, NLTK helps us convert raw text into a format that Machine Learning and Deep Learning models can work with.

**TextBlob** is a Python library used for simple Natural Language Processing (NLP) tasks. It provides an easy-to-use interface for processing and analyzing text without writing complicated NLP code.

Think of TextBlob as a beginner-friendly NLP toolkit, similar to NLTK, but with a simpler API.

**We are Using this for Correcting the Incorrect Words**

````python
import pandas as pd

data = {
    'text': [
        "I LOVE Machine Learning!!!",
        "Python is AMAZING for Data Science.",
        "The students are learning Deep Learning techniques.",
        "I don't like boring lectures...",
        "NLP is changing the world of AI!",
        "Running, walking and swimming are good exercises.",
        "The movie was absolutely fantastic!!!",
        "This product is NOT worth buying.",
        "Data Scientists analyse large amounts of data.",
        "He was running faster than the other players.",
        "I have been studying Natural Language Processing.",
        "The cats are playing in the beautiful gardens.",
        "AI, ML, and DL are interesting fields.",
        "Why is preprocessing so IMPORTANT?",
        "I am learning Python @ 2 AM!!! #coding"
    ]
}

df = pd.DataFrame(data)

print(df)
````

````python
#Converting text to LowerCase
df['text'] = df['text'].str.lower()
````

````python
# Removing URLs
import re
def remove_url(text):
  pattern=re.compile('https://www\.[A-Za-z]+\.com')
  return pattern.sub(r' ',text)
````

````python
df['text']=df['text'].apply(remove_url)
````

````python
# Removing Punctuations

punc = '!"#$%&\'()*+,-./:;<=>?@[\\]^_`{|}~'
print(punc)
````

````python
# def remove_punc(line):
#     for i in line:
#       if i in punc:
#         line=line.replace(i,'')

#     return line
def remove_punc(line):
  return line.translate(str.maketrans('','',punc))
````

````python
df['text']=df['text'].apply(remove_punc)
````

````python
# Correcting Words
#TextBlob is an open-source, beginner-friendly Python library used for processing textual data and performing natural language processing (NLP) tasks.
# Built on top of established NLP packages like NLTK (Natural Language Toolkit) and Pattern, TextBlob provides a simple, Pythonic API to handle common text-analysis operations without heavy boilerplate code
from textblob import TextBlob
````

````python
df['text']=df['text'].apply(lambda x : TextBlob(x).correct().string)
````

````python
# Removing Stop Words
import nltk
nltk.download('stopwords')

from nltk.corpus import stopwords
stop_words=stopwords.words('english')
````

````python
def remove_stopWords(line):
    words = line.split()

    for i in words:
        if i.lower() in stop_words:
            words.remove(i)

    return ' '.join(words)
````

````python
df['text']=df['text'].apply(remove_stopWords)
````

````python
#Removing WhiteSpaces from start and end
df['text']=df['text'].apply(str.strip)
````

````python
#Tokenization
#Tokenization is the process of breaking text into smaller units called tokens. These tokens can be words, subwords, characters or even bytes.

# For Tokenization We Can use Two Library - 1.NLTK and 2. Spacy

import nltk
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize,sent_tokenize

df['text']=df['text'].apply(word_tokenize)
````

**Corpus** – A corpus is a collection of text data used in NLP to train or analyse a model. It can contain thousands of sentences, reviews, articles, or documents. For example, if you are analysing 100 movie reviews, all 100 reviews together are called your corpus.

**Document** – A document is a single piece of text inside a corpus. It can be one sentence, one paragraph, or an entire article. For example, if your corpus contains 100 movie reviews, each individual review is called a document.

**Vocabulary** – A vocabulary is the collection of all unique words or tokens present in your corpus. It acts like a dictionary that a model uses to identify words and convert them into numerical representations. For example, if your dataset contains "I love pizza" and "I love pasta", the vocabulary will be ["I", "love", "pizza", "pasta"].


# Integer Encoding vs Bag of Words (BoW)

## 1. Integer Encoding

Integer Encoding assigns a unique integer to every word in the vocabulary.

Suppose we have two documents:

```python
D1 = "I love machine learning"
D2 = "I love deep learning"
```

First, create a vocabulary:

```python
vocabulary = {
    "I": 1,
    "love": 2,
    "machine": 3,
    "learning": 4,
    "deep": 5
}
```

Now convert each document into integer sequences:

| Document | Integer Sequence |
|----------|------------------|
| D1 | `[1, 2, 3, 4]` |
| D2 | `[1, 2, 5, 4]` |

**Important:** Integer Encoding preserves the order of words in a sentence.

## 2. Bag of Words (BoW)

Bag of Words represents each document using the frequency of words present in the vocabulary.

Using the same documents:

```python
D1 = "I love machine learning"
D2 = "I love deep learning"
```

First, create a vocabulary:

```python
vocabulary = ["I", "love", "machine", "learning", "deep"]
```

Now count how many times each word appears in each document.

| Document | I | love | machine | learning | deep |
|----------|---|------|---------|----------|------|
| D1 | 1 | 1 | 1 | 1 | 0 |
| D2 | 1 | 1 | 0 | 1 | 1 |

Therefore, the numerical representations become:

```python
D1 = [1, 1, 1, 1, 0]
D2 = [1, 1, 0, 1, 1]
```

**Important:** BoW does not preserve the order of words. It only considers their frequency.

## 3. Visual Difference

**Original Text:**

D1: I love machine learning

D2: I love deep learning

**Integer Encoding:**

`[1, 2, 3, 4]`

One ID per word, preserving order.

**Bag of Words:**

`[1, 1, 1, 1, 0]`

One frequency vector per document.

## 4. Major Differences

| Feature | Integer Encoding | Bag of Words |
|---------|------------------|--------------|
| Representation | Integer IDs | Frequency vector |
| Output | Sequence of integers | Vector per document |
| Word order | Preserved | Ignored |
| Word frequency | Not directly represented | Represented |
| Vocabulary | Required | Required |
| Typical usage | RNN, LSTM, Embedding | Traditional ML |
| Semantic meaning | No inherent meaning in IDs | No inherent meaning in counts |


# N-Grams in NLP

## 1. What are N-Grams?

N-Grams are continuous sequences of N items (usually words or tokens) extracted from a text.

They are used in NLP to capture the relationship and order between neighbouring words.

Unlike Bag of Words (BoW), which only considers word frequency and ignores word order, N-Grams can capture short sequences of consecutive words.

For example:

```python
text = "I love machine learning"
```

Depending on the value of N, we can generate different N-Grams.

## 2. Types of N-Grams

There are three commonly used types of N-Grams:

### A. Unigram (N = 1)

A Unigram considers one word at a time.

Example:

```python
text = "I love machine learning"
```

Output:

```python
["I", "love", "machine", "learning"]
```

Here, each individual word is treated as a separate N-Gram.

### B. Bigram (N = 2)

A Bigram considers two consecutive words at a time.

Example:

```python
text = "I love machine learning"
```

Output:

```python
[
    "I love",
    "love machine",
    "machine learning"
]
```

Here, every N-Gram contains two consecutive words.

### C. Trigram (N = 3)

A Trigram considers three consecutive words at a time.

Example:

```python
text = "I love machine learning"
```

Output:

```python
[
    "I love machine",
    "love machine learning"
]
```

Here, every N-Gram contains three consecutive words.

## 3. Visual Representation

Suppose we have:

"I love machine learning"

| N-Gram | Result |
|--------|--------|
| Unigram (N=1) | I, love, machine, learning |
| Bigram (N=2) | I love, love machine, machine learning |
| Trigram (N=3) | I love machine, love machine learning |
| 4-Gram (N=4) | I love machine learning |

## 4. How Many N-Grams Are Generated?

For a sentence containing L tokens, the number of N-Grams of size N is:

Number of N-Grams = L - N + 1

For example:

Sentence = "I love machine learning"

Number of tokens (L) = 4

Unigram (N=1):
4 - 1 + 1 = 4

Bigram (N=2):
4 - 2 + 1 = 3

Trigram (N=3):
4 - 3 + 1 = 2

4-Gram (N=4):
4 - 4 + 1 = 1

## 5. Difference Between BoW and N-Grams

Bag of Words considers individual word frequencies, whereas N-Grams can consider consecutive word sequences.

Example:

D1 = "I love Python"
D2 = "Python love I"

Using Unigram (N=1), both documents contain the same words with the same frequencies.

However, using Bigram (N=2):

D1:
- I love
- love Python

D2:
- Python love
- love I

The Bigram representations will be different.

| Feature | Bag of Words | N-Grams |
|---------|--------------|---------|
| Word frequency | Yes | Yes |
| Word order | Ignored | Partially captured |
| Individual words | Yes | Yes, if N=1 |
| Consecutive word sequences | No | Yes |
| Vocabulary size | Usually smaller | Can become larger |

Note: Unigram is equivalent to the word-level Bag of Words approach when both use the same vocabulary and counting method.

## 8. Advantages of N-Grams

1. Captures local word order.
2. Helps identify common phrases.
3. Can improve text classification.
4. Helps capture expressions such as "not good", "very bad" and "highly recommended".
5. Can be used in traditional ML models such as Naive Bayes and Logistic Regression.

## 9. Disadvantages of N-Grams

1. Vocabulary size increases as N increases.
2. Requires more memory and computation.
3. Many possible N-Grams may occur very rarely.
4. Does not capture long-distance relationships between words.
5. Does not understand the actual semantic meaning of words.

- Next Word Prediction
- Search Engines
- Text Mining


# TF-IDF (Term Frequency – Inverse Document Frequency)

## 1. What is TF-IDF?

TF-IDF stands for Term Frequency – Inverse Document Frequency.

It is a text feature extraction technique used in NLP to measure the importance of a word in a particular document relative to the entire corpus.

Unlike Bag of Words (BoW), which only counts word occurrences, TF-IDF gives higher importance to words that are frequent in a particular document but less frequent across the entire corpus.

The main idea is:

- Words appearing frequently in a document can be important.
- Words appearing in almost every document are generally less informative.
- Words appearing in fewer documents can be more useful for distinguishing documents.

## 2. Why do we need TF-IDF?

Suppose we have three documents:

```python
D1 = "I love machine learning"
D2 = "I love deep learning"
D3 = "I love Python programming"
```

Notice that the words "I" and "love" appear in all three documents.

These words do not help us distinguish one document from another.

However, words like "machine", "deep" and "Python" are more specific to individual documents.

TF-IDF helps assign lower importance to common words and higher importance to more distinctive words.

## 3. Components of TF-IDF

TF-IDF consists of two components:

### A. Term Frequency (TF)

Term Frequency measures how frequently a term appears in a particular document.

Formula:

TF(t, d) = Number of times term t appears in document d / Total number of terms in document d

Example:

```python
text = "I love machine learning machine"
```

Total words = 5

Frequency of "machine" = 2

TF(machine) = 2 / 5 = 0.4

Therefore, the Term Frequency of "machine" is 0.4.

### B. Inverse Document Frequency (IDF)

Inverse Document Frequency measures how rare or informative a term is across the entire corpus.

A word appearing in almost every document gets a lower IDF score.

A word appearing in fewer documents gets a higher IDF score.

Formula:

IDF(t) = log(N / DF(t))

Where:

- N = Total number of documents in the corpus.
- DF(t) = Number of documents containing term t.
- log = Logarithmic function.

Example:

Suppose we have 3 documents:

```python
D1 = "I love machine learning"
D2 = "I love deep learning"
D3 = "I love Python programming"
```

Total documents (N) = 3

The word "love" appears in all 3 documents.

IDF(love) = log(3 / 3) = 0

The word "machine" appears in only 1 document.

IDF(machine) = log(3 / 1)

Therefore, "machine" receives a higher IDF score than "love".

Note: This is the basic IDF formula. Scikit-learn uses a smoothed IDF formula by default.

### C. TF-IDF

TF-IDF is calculated by multiplying Term Frequency and Inverse Document Frequency.

Formula:

TF-IDF(t, d) = TF(t, d) × IDF(t)

A higher TF-IDF score generally indicates that a word is more important for distinguishing a particular document within the corpus.

## 4. TF-IDF vs Bag of Words

| Feature | Bag of Words | TF-IDF |
|---------|--------------|--------|
| Purpose | Counts word occurrences | Measures word importance |
| Frequency | Uses word counts | Uses TF and IDF |
| Common words | Can receive high counts | Generally receive lower weights |
| Rare words | No special importance adjustment | Can receive higher weights |
| Output | Frequency vector | Weighted frequency vector |
| Representation | Sparse | Sparse |
| Typical usage | Traditional ML | Traditional ML |

## 6. Important Parameters of TfidfVectorizer

```python
TfidfVectorizer(
    lowercase=True,
    stop_words=None,
    ngram_range=(1, 1),
    max_features=None,
    max_df=1.0,
    min_df=1,

    smooth_idf=True,
```

Important parameters:

- lowercase: Converts text into lowercase.
- stop_words: Removes stopwords.
- ngram_range: Specifies which N-Grams to generate.
- max_features: Limits the vocabulary size.
- max_df: Ignores terms appearing in too many documents.
- min_df: Ignores terms appearing in too few documents.
- use_idf: Enables or disables IDF weighting.
- smooth_idf: Adds smoothing to IDF calculation.
- sublinear_tf: Applies logarithmic scaling to term frequency.

# Sparse vs Dense Representations in NLP

## 1. What is a Sparse Representation?

A Sparse Representation is a numerical representation in which most of the values are zero.

In NLP, techniques like Bag of Words (BoW), N-Grams and TF-IDF generally produce sparse representations.

This happens because the vocabulary can contain thousands of unique words, but an individual document usually contains only a small fraction of those words.

### Example

Suppose we have the following vocabulary:

```python
vocabulary = ["I", "love", "machine", "learning", "deep", "Python"]
```

Document:

```python
text = "I love machine learning"
```

Its Bag of Words representation will be:

```python
[1, 1, 1, 1, 0, 0]
```

Notice that most values are zero because the document does not contain every word in the vocabulary.

This is called a Sparse Representation.

## 2. What is a Dense (Non-Sparse) Representation?

A Dense Representation is a numerical representation in which most values are non-zero.

In NLP, Word Embeddings such as Word2Vec, GloVe and neural network Embedding Layers produce dense representations.

Instead of representing words using a large vector containing mostly zeros, embeddings represent words using smaller, dense numerical vectors.

### Example

Suppose the word "machine" is represented using a 4-dimensional embedding:

```python
machine = [0.25, 0.67, 0.12, 0.89]
```

Notice that most or all values are non-zero.

This is called a Dense Representation.

Note: These values are illustrative. Actual embeddings are learned during training or obtained from pretrained models.

## 3. Sparse vs Dense Representation

| Feature | Sparse Representation | Dense Representation |
|---------|------------------------|----------------------|
| Values | Mostly zeros | Mostly non-zero |
| Dimensions | Often very high | Usually lower |
| Memory | Can be inefficient when stored as a dense array | Compact for a fixed number of dimensions |
| Semantic relationships | Not inherently captured | Can learn semantic relationships |
| Examples | BoW, N-Grams, TF-IDF | Word2Vec, GloVe, Embeddings |
| Typical usage | Traditional Machine Learning | Deep Learning |

## 4. Visual Difference

### Sparse Vector

```python
[1, 0, 0, 0, 1, 0, 0, 0, 1, 0]
```

Most values are zero.

### Dense Vector

```python
[0.21, 0.45, 0.67, 0.12, 0.89]
```

Most values are non-zero.

## 5. Why are BoW and TF-IDF Sparse?

Suppose our vocabulary contains 10,000 unique words.

An individual document may contain only 100 unique words from that vocabulary.

Therefore, most of the 10,000 positions in its feature vector will contain zero.

This creates a Sparse Representation.

## 6. Why are Word Embeddings Dense?

Word Embeddings represent words using a smaller number of dimensions, such as 50, 100, 300 or 768.

Each dimension contains a learned numerical value.

These values allow neural networks to learn useful relationships between words.

For example, embeddings may learn that:

- King and Queen are related.
- Cat and Dog are related.
- Happy and Joyful are related.

## 8. Important Takeaways

1. Sparse representations contain mostly zero values.
2. Dense representations contain mostly non-zero values.
3. BoW, N-Grams and TF-IDF generally produce sparse representations.
4. Word2Vec, GloVe and Embedding Layers produce dense representations.
5. Sparse representations often have high dimensionality.
6. Dense embeddings can learn semantic relationships between words.
7. Sparse matrices can save memory when stored using suitable sparse formats.
