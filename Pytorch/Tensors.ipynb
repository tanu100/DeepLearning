# **Tensor**
A tensor is a general-purpose data structure used to store numerical data in multiple dimensions. It is the fundamental building block of Deep Learning frameworks like PyTorch and TensorFlow.

Think of it this way:
- Scalar → Single number.
- Vector → One-dimensional collection.
- Matrix → Two-dimensional collection.
- Tensor → Generalized multidimensional collection.

Syntax-


torch.tensor(data, dtype=None, device=None)

#**0D Tensor – Scalar**

A single numerical value.

```python
import torch
```

```python
x=torch.tensor(5)
print(x)
print(x.ndim)
print(x.shape)
```

Explanation:
- ndim tells you how many dimensions exist.
- shape tells you the size along each dimension.
- A scalar has no axes, so its shape is empty.

# **1D Tensor – Vector**

```python
x = torch.tensor([10, 20, 30, 40])

print(x.ndim)
print(x.shape)
print(x.numel()) # No of Elements
```

# **2D Tensor – Matrix**

```python
x = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])

print(x.ndim)
print(x.shape)
print(x.dtype)
print(x.numel()) # No of Elements
```

# **3D Tensor**

```python
x = torch.tensor([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print(x.ndim)
print(x.shape)

print(x.device) #Tells Us About the Processing Unit - GPU Or CPU

print(x.numel()) # No of Elements
```

```python
# Checking if GPU is enable or not
if torch.cuda.is_available():
  print("GPU is enabled")
else:
  print("CPU is enabled")
```

```python
device = 'cuda' if torch.cuda.is_available() else 'cpu'
x = torch.tensor([
    [
        [1,2,3],
        [4,5,6]
    ]
],dtype=torch.float32,device=device)
print(x.ndim)
print(x.shape)
print(x.dtype)
```

# **Functions Within Tensors**

# **torch.zeros()**

– Create tensors filled with zeros

```python
x=torch.zeros(5)
print(x)
```

```python
x=torch.zeros((2,3)) # multidimensional
print(x)
print(x.dtype)
```

```python
x=torch.zeros((3,3),dtype=torch.int64)
print(x)
print(x.dtype)
```

# **torch.ones()**

– Create tensors filled with ones

```python
x=torch.ones(5)
print(x)
```

```python
x = torch.ones(3, 4)
print(x)
print(x.dtype)
```

```python
x = torch.ones((3,3),dtype=torch.int64)
print(x)
print(x.dtype)
```

# **torch.randn()**

Random numbers from a normal distribution

It generates random numbers from a standard normal distribution.

A standard normal distribution has:
- Mean = 0
- Standard deviation = 1

```python
x=torch.randn(5)
print(x)
```

```python
x=torch.randn((3,3))
print(x)
```

# **torch.rand()**

- Random numbers from a uniform distribution
Another random tensor creation function.


  Unlike randn(), it generates values uniformly between 0 and 1.

```python
x=torch.rand(5)
print(x)
```

```python
x=torch.rand((3,3))
print(x)
```

# **torch.arange()**
Generate a sequence of numbers

torch.arange(start, end, step)


```python
x=torch.arange(0,10,1)
print(x)
```

# **torch.linspace()**

Generate evenly spaced numbers

torch.linspace(start, end, no_of_elements_between_start_and_end,dtype = ?)

```python
x=torch.linspace(0,10,5)
print(x)
```

```python
x=torch.linspace(0,10,2,dtype=torch.int64)
print(x)
```

# **torch.empty_like()**

 – Uninitialized tensor
This one is slightly different.


It creates a tensor with the same shape as the input, but does not initialize its values.

```python
torch.empty_like(x)
```

These values are unpredictable. They are whatever data happens to be present in the allocated memory.

They are not guaranteed to be zeros or random numbers.

# **torch.zeros_like()**

Fill with zeros


Creates a new tensor with the same shape as the input tensor, filled with zeros.

```python
torch.zeros_like(x)
```

# **torch.ones_like()**

– Fill with ones


Creates a new tensor with the same shape, filled with ones.

```python
torch.ones_like(x)
```

# **torch.rand_like()**

 – Random uniform values


Creates a tensor with the same shape as the input, filled with random values from a uniform distribution between 0 and 1.

```python
torch.rand_like(x,dtype=torch.float32)
```

# **Mathematical Operations**

```python
# Scalar Operation

x=torch.full((3,3),5)

# Addition
x + 5
# Subtraction
x-5
# Mulitple
x * 5
#Divide
x / 5
# power
x ** 2
# mod
x % 5
```

```python

x=torch.full((3,3),5)
y=torch.full((3,3),5)

# Addition
x + y
# Subtraction
x-y
# Mulitple
x * y
#Divide
x / y
# power
x ** y
# mod
x % y
```

```python
# SHAPES MUST BE SAME FOR THE OPERATION
x=torch.full((2,3),5)
y=torch.full((3,3),5)

# Addition
x + y
# Subtraction
x-y
# Mulitple
x * y
#Divide
x / y
# power
x ** y
# mod
x % y
```

```python
# torch.abs(tensor) - gives positive value {Absolute}

z = torch.tensor([-2,10,-19,5])

print(torch.abs(z))
```

Reduction Operation in Maths

```python
x = torch.randint(size=(2,3),low=0,high=10,dtype=torch.float32)
print(x)
```

```python
#Sum
torch.sum(x)
#Sum along Columns
torch.sum(x ,dim=0)
#Sum along rows
torch.sum(x,dim=1)
```

```python
#Mean - Only Works with Float or Long
torch.mean(x)
#Mean  along col
torch.mean(x,dim=0)
```

```python
#Median
torch.median(x)
```

```python
#Max
torch.max(x)
torch.min(x)
```

```python
#argmin - position of minimum element in matrix
torch.argmin(x)
#argmax - position of maximum element in matrix
torch.argmax(x)
```

# Matrix Operations

```python
print(torch.matmul(A,B))
```

```python
print(torch.transpose(A,0,1))

# Better Way
print(A.T)
```

# **Copying a Tensor**

```python
a=torch.randn((3,3))
print(a)
b=a
print(b)
print("id of a is ",id(a))
print("id of b is ",id(b))
```

```python
a[0][0]=0
print(a)
print(b)
```

```python
c=torch.clone(a)
a[0][0]=100
print(a)
print(b)
print(c)
```
