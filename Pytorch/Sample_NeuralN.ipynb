```python
import torch
import torch.nn as nn
```

```python
class DemoNN(nn.Module):

    def __init__(self,num_features):

        super().__init__()
        
        self.network= nn.Sequential(
            nn.Linear(num_features,64), # network.0
            nn.ReLU(), # network.1
            nn.Linear(64,32), # network.2
            nn.ReLU(), # network.3
            nn.Linear(32,1), # network.4
            nn.Identity()  # network.5
        )


    def forward(self,features):

        out=self.network(features)

        return out
    
        
```

```python
from sklearn.datasets import make_regression

from sklearn.model_selection import train_test_split
```

```python
x,y=make_regression(noise=0.3,n_features=3,n_targets=1,random_state=42,n_samples=1000)

X_train,x_test,Y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.3)
```

```python
#Testing only the Forward Pass
model=DemoNN(X_train.shape[1])

TX_train=torch.from_numpy(X_train).float()
TY_train=torch.from_numpy(Y_train).float().reshape(-1,1)

TX_test=torch.from_numpy(x_test).float()
TY_test=torch.from_numpy(y_test).float().reshape(-1,1)
```

```python
#Loss Function Intislisation
loss_fn=nn.MSELoss()

optimizer=torch.optim.Adam(
    model.parameters(),
    lr=0.01
)
```

```python

model.train()

y_pred=model(TX_train)

loss=loss_fn(y_pred,TY_train)

optimizer.zero_grad()
loss.backward()

for name, param in model.named_parameters():
    print(name, param.grad)

# 10. Update weights
optimizer.step()

print("Loss:", loss.item())
```

```python
for param in model.parameters():
    print(param)
```

```python
# Switch to evaluation mode
model.eval()

# Disable gradient calculations
with torch.no_grad():

    # Predictions on test data
    y_pred = model(TX_test)

    # Calculate test MSE
    test_loss = loss_fn(y_pred, TY_test)

print("Test MSE:", test_loss.item())
```

```python

epochs = 200

for epoch in range(epochs):

    # 1. Training mode
    model.train()

    # 2. Forward propagation
    train_pred = model(TX_train)

    # 3. Calculate training loss
    train_loss = loss_fn(train_pred, TY_train)

    # 4. Backpropagation and weight update
    optimizer.zero_grad()
    train_loss.backward()
    optimizer.step()

    # 5. Evaluation mode
    model.eval()

    # 6. Calculate test loss without gradients
    with torch.no_grad():
        test_pred = model(TX_test)
        test_loss = loss_fn(test_pred, TY_test)

    if (epoch + 1) % 20 == 0:
        print(
            f"Epoch [{epoch+1}/{epochs}] "
            f"Train MSE: {train_loss.item():.4f} "
            f"Test MSE: {test_loss.item():.4f}"
        )
```

```python
# Switch to evaluation mode
model.eval()

# Disable gradient calculations
with torch.no_grad():

    # Predictions on test data
    y_pred = model(TX_test)

    # Calculate test MSE
    test_loss = loss_fn(y_pred, TY_test)

print("Test MSE:", test_loss.item())
```
